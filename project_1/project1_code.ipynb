### Step 3: The Hard Way!

In this step I will calculate the same statistics without using pandas module. First I will use python's build in module csv to read through the csv file, and saving the column of interest in two lists.

```python
import csv

with open("taiwan_bankruptcy.csv", newline="") as file:
    rows = list(csv.DictReader(file))

    bankrupt_values = []
    debt_ratio_values = []

    for row in rows:
        bankrupt_values.append(int(row["Bankrupt?"]))
        debt_ratio_values.append(float(row[" Debt ratio %"]))
```

Now we can calculate mean, median and mode of the two lists. To make it more interesting, I decided to define functions for this step.

```python
# function for mean
def mean(lst):
    n = len(lst)
    return sum(lst) / n


# function for median
def median(lst):
    n = len(lst)
    ordered = sorted(lst)
    middle = n // 2

    # my though here is that using // gives a number that is rounded down, since first item in list is indexed 0. For a list with odd
    # number of items, n//2 points directly to the median.

    # for a list with odd number of items
    if n % 2 == 1:
        return ordered[middle]
    # for a list with even number of items
    else:
        return (ordered[middle - 1] + ordered[middle]) / 2
        # here is middle - 1 because middle value of an even list points at the right of the two value since python start index with 0.


# function for mode
def mode(lst):
    # We can create a dictionary, with key being the item in the list, and value being how many times they appeared in the list
    # if an item is not in the dictionary, we take that item as key and record value as 1
    # if we have seen this item, we simple add the value of that key by 1.
    counts = {}

    for item in lst:
        # this part handles items that's already in counts
        if item in counts:
            counts[item] = counts[item] + 1

        # this part handles items that's not in counts
        else:
            counts[item] = 1

        # now we return mode as a list. The modes are keys in the dictionary with the highest value.

    highest_count = max(counts.values())
    modes = []

    for item in counts:
        if counts[item] == highest_count:
            modes.append(item)

    return modes
```

Now that we have defined the functions, we can apply them to find the mean, median, and mode of each list.

```python
# bankruptcy list

print("STATISTICS OF BANKRUPTCY")
print("Mean: ", mean(bankrupt_values))
print("Median:", median(bankrupt_values))
print("Modes:", mode(bankrupt_values))
```

```python
# Debt ratio list

print("STATISTICS OF Debt Ratio")
print("Mean: ", mean(debt_ratio_values))
print("Median:", median(debt_ratio_values))
print("Modes:", mode(debt_ratio_values))
```

### Step 4: Data Visualization

To visualize data, I want to use the dummy variable 'Bankrupt' to group the debt ratios for all the bankrupted companies and all the companies that are normally operating. In this way, I can learn whether debt ratios signals unhealthy operating condition.

```python
# First find the median debt ratios of the two groups
medians = bankruptcy_subset.groupby("Bankrupt?")[" Debt ratio %"].median()

print("Median debt ratio by bankruptcy status")
print("Each # represents about 0.01 on the debt ratio scale.\n")

# since we can't use plotly. A nice visualization will be a horizontal bar plot representing with '#'. Each '#' represent 0.01% debt ratio.
for status, label in [(0, "Not bankrupt"), (1, "Bankrupt")]:
    value = medians[status]
    bar = "#" * round(value / 0.01)
    print(f"{label:12} | {bar} {value:.4f}")
```

The median debt ratio was higher among observations labeled bankrupt (0.1874) than among those labeled not bankrupt (0.1094).
